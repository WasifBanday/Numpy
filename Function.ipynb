                                                        STACKING

```python
a4=np.arange(12).reshape(3,4)
a5=np.arange(12).reshape(3,4)
```

```python
#      horizontal stacking ---> Joins them horizontally
np.hstack((a4,a5))
```

```python
#      vertical stacking ---> Joins them vertically
np.vstack((a4,a5))
```

                                                SPLITTING--> Splits two arrays

```python
np.hsplit(a4,2)  # a4 array ko 2 equal parts mai split krdo horizontally
np.vsplit(a4,3)
```
