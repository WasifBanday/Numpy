```python
import numpy as np
```

                                                                     INDEXING  --- only one item can be fetched 

```python
a1=np.arange(10)
a2=np.arange(12).reshape(3,4)
a3=np.arange(27).reshape(3,3,3)
```

```python
#   1D

print(a1)
a1[-1]    # Fetch last element 
```

```python
#  2D

print(a2)
a2[1,2]   # 1--for(row)       2---for(column)
```

```python
#  3D --  it's a combination of two 2D

print(a3)
a3[1,0,1]   #  1--for(which 2D)     # 0--for(row no of that 2D)   # 1 --for(col no of that 2D)
```

                                                                          SLICING -- Can fetch multiple items

                                                                            1D

```python
print(a1)
a1[2:5]
```

                                                                            2D

```python
print(a2)
```

```python
a2[:,2]   # row sarey fetch krdo , column only 3rd wla karo
```

```python
a2[1:,1:3]    # from fow 1st index to all , from col 1st index to col 3rd-1
```

```python
a2[::2,::3]   #  ::2 --> All alternate rows , ::3 --> Takes every 3rd col
```

                                                                            3D

```python
a3
```

```python
a3[1]  # Prints whole 2nd 2D array
```

```python
a3[::2]  # prints 1st and last 2D array
```

```python
a3[0,1,:]   # prints 2nd row of 1st 2D array
```

```python
a3[1,:,1]   # prints 2nd col of 2nd 2D array
```

```python
a3[::2,0,::2]   
```
