```python
import numpy as np
```

```python
# 1D
a1=np.arange(10)
for i in a1:
    print(i)
```

```python
# 2D
a2=np.arange(12).reshape(3,4)
for i in a2 :
    print(i)              # prints each row at a time
```

```python
#  3D
a3=np.arange(27).reshape(3,3,3)
for i in a3 :
    print(i)             # prints each 2D array at a time
```

```python
for i in np.nditer(a3):
    print(i)
```
