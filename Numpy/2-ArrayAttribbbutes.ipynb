```python
import numpy as np
```

```python
a1=np.arange(10)   # 1D
a2=np.arange(12,dtype=float).reshape(3,4)   # 2D
a3=np.arange(8).reshape(2,2,2)   # 3D
```

```python
# ndim : Gives us no. of dimensions
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# Shape : Gives us No. of rows and cols
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
# Size : Gives us no. of items in an array
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# Itemsize : Gives us size or each item occupied in memory
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
# dtype : Tells us Data-Type
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
# astype : Changes dataType

a3.astype(np.int32)
```
