```python
import numpy as np 
```

```python

a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
# 1 : min / max / sum /prod

print(np.min(a1))
print(np.max(a1))
print(np.sum(a1))
print(np.prod(a1))
```

```python
# We can also find max / min etc  , of each row and column 
#  0 ---> for Column     ,,,,, 1 ---> for Row 

print(np.max(a1,axis=1))
print(np.min(a1,axis=0))
```

```python
# 2 : mean / median / std / var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
```

Dot Product --> Column of 1st array should be equal to Row of second array

```python
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
```

```python
np.dot(a3,a2)
```

Log and Exponent 

```python
print(np.log(a1))
print(np.exp(a1))
```

Round , Floar , ceil 

```python
print(np.round(np.random.random((2,3))*100))
print(np.floor(np.random.random((2,3))*100))
print(np.ceil(np.random.random((2,3))*100))
```
