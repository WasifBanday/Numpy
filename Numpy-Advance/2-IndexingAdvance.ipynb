```python
import numpy as np
```

Two types of indexing is there 
    1-Fancy Indexing 
    2-Boolean Indexing 

                                                    1-Fancy Indexing 

```python
a=np.arange(24).reshape(6,4)
print(a)
```

```python
# Question - I want output 1st , 3rd , 4th and 6th row 
a[[0,2,3,5]]
```

```python
# Question - I want output 1st , 3rd and 4th column 
a[:,[0,2,3]]
```

                                                    2-Boolean Indexing 


```python
a=np.random.randint(1,100,24).reshape(6,4)    # randint()  , generates numbers randomly with given range { Start(1) - To(100) - no of items(24) }
print(a)
```
