```python
import numpy as np
```

Two types of indexing is there 
    1-Fancy Indexing 
    2-Boolean Indexing 

                                                    1-Fancy Indexing 

```python
a=np.arange(24).reshape(6,4)
print(a)
```

```python
# Question - I want output 1st , 3rd , 4th and 6th row 

a[[0,2,3,5]]
```

```python
# Question - I want output 1st , 3rd and 4th column 

a[:,[0,2,3]]
```

                                                    2-Boolean Indexing 


```python
a=np.random.randint(1,100,24).reshape(6,4)    # randint() , generate numbers randomly within given range { Start(1) _ To(100) _ no of items(24) }
print(a)
```

```python
# Question - Find All numbers greater then 50

a>50 # This will give us boolean result 
a[a>50]  # This will give us numbers which are true 
```

```python
# Question - Find out even numbers 

a % 2 == 0    # This will give us boolean result 
a[a % 2 == 0] # This will give us numbers which are true 
```

```python
# Question - Find All numbers greater then 50 and are even

(a > 50) & (a % 2 == 0)
a[(a > 50) & (a % 2 == 0)]
```

```python
# Question - Find All numbers which are not divisible by 7

a[~(a%7==0)]
```
